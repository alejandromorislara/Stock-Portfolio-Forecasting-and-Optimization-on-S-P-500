### **Merging Stock data and Macroeconomic data**

After obtaining individual CSV files containing data for each stock, our next step was to construct a unified, comprehensive dataset. Initially, we considered an approach where each company’s data would be arranged in contiguous blocks, with repeated date entries for each firm and only six columns : *Date*, *Close*, *AdjustedClose*, *Open*, *High*, and *Volume* in the original_data dataframe. While this method would keep the dataset structure relatively simple, we identified several drawbacks. First, this vertical stacking of time-series data for each company could prevent the model from capturing important **inter-company relationships**. Moreover, implementing such an arrangement would complicate the programming process and significantly increase computational demands.

To address these challenges, we opted to merge the CSV files for each stock **horizontally**, aligning their data by dates. In other words, we combined the datasets as additional columns rather than stacking them by rows. This horizontal integration enables the model to simultaneously analyze the time-series behavior of multiple stocks, thereby facilitating the discovery of cross-company correlations and interdependencies. By leveraging a richer information structure, the model is better positioned to uncover complex patterns and improve predictive accuracy. Ultimately, this approach not only enhances the model’s explanatory power but also reduces the computational overhead associated with more convoluted dataset configurations. 

#### **Date range challenge**

This strategy ensures that the dataset remains consistent across all constituent stocks, enabling more straightforward modeling and analysis of time-series relationships within the S&P 500 context.

Furthermore, not all stocks commenced trading on the same initial date. To address potential gaps at the start of each series gaps that would otherwise generate numerous NaN entries we employed a data-filling technique. For dates lacking stock-specific information, we replaced missing values with **zeros**. While this approach introduces a neutral placeholder rather than an imputed economic value, it serves to preserve the shape and integrity of the dataset. This, in turn, facilitates subsequent modeling efforts by ensuring uniform input dimensions, reducing the need for complex imputations, and allowing the learning algorithm to focus on detecting meaningful economic and market-driven patterns rather than being confounded by irregular data structures.

### **Difference between Adjusted Close and Close** :

The Close price of a stock represents the last trading price recorded at the end of the market session for a specific date. However, it does not account for corporate events such as dividends, stock splits, mergers, acquisitions, spin-offs, or other actions that can alter the company’s capital structure and historical price records. In contrast, the Adjusted Close price systematically incorporates these adjustments, enabling a more consistent and comparable view of the stock's value over time.

By using the Adjusted Close price, the time series becomes free from arbitrary discontinuities caused by structural changes to the stock. For instance, in the case of a 2:1 stock split, the nominal price is halved without affecting the company's fundamental value. This adjustment ensures that historical analysis provides a more accurate representation of the stock's underlying performance, eliminating artificial fluctuations and facilitating the identification of genuine trends.

Machine learning and deep learning models often rely on stable and consistent patterns within time series data to make reliable predictions. The use of Adjusted Close prices reduces the "noise" introduced by corporate events that do not reflect changes in the stock's intrinsic economic value. This leads to a more accurate extraction of meaningful market signals. Such adjustments are particularly critical for models with medium- or long-term prediction horizons, as they enhance the reliability and interpretability of the model’s output by focusing on real trends rather than transient, non-economic artifacts in the data.

### **Obtaining Technical Indicators**

Tras añadir los datos financieros y macroeconómicos, vamos a obtener una serie de indicadores técincos para los datos de los stock a través de la librería **ta**, la cual se trata de una librería para realizar análisis de series temporales. Gracias a esta librería hemos seleccionado una serie de características tomando como referencia el siguiente paper [Bao W, Yue J, Rao Y. A deep learning framework for financial time series using stacked autoencoders and long-short term memory.](https://pubmed.ncbi.nlm.nih.gov/28708865/). Los indicadores técnicos que hemos seleccionado han sido los siguientes : 



Subscripts $t$ indicate the current time step, and $n$ denotes the chosen window length.

1. **Moving Average Convergence Divergence (MACD)**  
MACD is a momentum-based trend indicator derived from the difference between two Exponential Moving Averages (EMAs) of the closing price. It helps identify potential trend reversals and measure the strength of current market momentum
$$
\text{MACD} = \text{EMA}(\text{Close}, n_{\text{short}}) - \text{EMA}(\text{Close}, n_{\text{long}})
$$

- Common parameters: $n_{\text{short}} = 12$, $n_{\text{long}} = 26$.

2. **Commodity Channel Index (CCI)**  
The CCI measures the deviation of a security’s price from its average price level, indicating whether it is overpriced or underpriced relative to its typical price range. It aids in identifying cyclical trends and potential overbought/oversold market conditions.

$$
\text{Typical Price} = \frac{\text{High} + \text{Low} + \text{Close}}{3}
$$

$$
\text{CCI} = \frac{\text{Typical Price} - \text{SMA}(\text{Typical Price}, n)}{0.015 \cdot \text{Mean Deviation}}
$$

3. **Average True Range (ATR)**  
ATR measures market volatility without indicating price direction. It estimates the degree of price movement and is commonly used in risk management.

$$
\text{TR}_t = \max\{\text{High}_t - \text{Low}_t, |\text{High}_t - \text{Close}_{t-1}|, |\text{Low}_t - \text{Close}_{t-1}|\}
$$

$$
\text{ATR} = \text{MA}(\text{TR}, n)
$$

- Common window: $n = 14$.

4. **Bollinger Bands (BOLL_upper, BOLL_lower)**  
Bollinger Bands consist of a moving average (middle band) of the closing price, and two bands set at a chosen number of standard deviations above and below the mean. They dynamically expand and contract with market volatility, helping identify potential price breakouts.
$$
\text{BOLL\_middle} = \text{MA}(\text{Close}, n)
$$

$$
\text{BOLL\_upper} = \text{BOLL\_middle} + k \cdot \sigma(\text{Close}, n)
$$

$$
\text{BOLL\_lower} = \text{BOLL\_middle} - k \cdot \sigma(\text{Close}, n)
$$

- Common parameters: $n = 20$, $k = 2$.

5. **Exponential Moving Average (EMA20)**  
The EMA assigns a higher weight to recent data points, making it more responsive than a Simple Moving Average.

$$
\text{EMA}(\text{Close}, n) = \alpha \cdot \text{Close}_t + (1 - \alpha) \cdot \text{EMA}(\text{Close}, n)_{t-1}
$$

$$
\alpha = \frac{2}{n+1}
$$

- Common window: $n = 20$.

6. **Simple Moving Averages (MA5, MA10)**  
Simple Moving Averages smooth out short-term fluctuations by averaging a series of closing prices over the last $n$ periods.

$$
\text{MA}_n = \frac{1}{n}\sum_{i=0}^{n-1} \text{Close}_{t-i}
$$

- For MA5 and MA10, $n = 5$ and $n = 10$, respectively.

7. **Momentum (MTM6, MTM12)**  
Momentum measures the rate of price change over a defined period, indicating whether prices are accelerating or decelerating.

$$
\text{MTM}_n = \frac{\text{Close}_t - \text{Close}_{t-n}}{\text{Close}_{t-n}}
$$

- For MTM6 and MTM12, $n = 6$ and $n = 12$, respectively.

8. **Rate of Change (ROC)**  
ROC calculates the percentage change in a security’s closing price over a specified time interval.

$$
\text{ROC} = \frac{\text{Close}_t - \text{Close}_{t-n}}{\text{Close}_{t-n}} \times 100
$$

9. **Stochastic Momentum Index (SMI)**  
The SMI refines the traditional Stochastic Oscillator by centering the price relative to the midpoint of the highest high and lowest low over a given period.

$$
\text{SMI} = \frac{\text{Close}_t - \frac{\text{Highest High} + \text{Lowest Low}}{2}}{(\text{Highest High} - \text{Lowest Low})/2} \times 100
$$

10. **Williams Volume Accumulation/Distribution (WVAD)**  
WVAD integrates both price movement and trading volume to evaluate whether price trends are supported by sufficient market participation.

$$
\text{WVAD} = \frac{\text{Close}_t - \text{Open}_t}{\text{High}_t - \text{Low}_t} \times \text{Volume}_t
$$

11. **Relative Strength Index (RSI)**  
RSI compares recent price gains versus losses over a chosen period, identifying overbought (RSI > 70) or oversold (RSI < 30) conditions.

$$
\text{RSI} = 100 - \frac{100}{1 + RS}
$$

$$
RS = \frac{\text{Average Gain}}{\text{Average Loss}}
$$

By incorporating these indicators, the dataset captures a multidimensional view of market dynamics, including trends, momentum, volatility, volume, and overbought/oversold conditions. This enhanced feature set improves predictive capabilities in quantitative modeling frameworks.

#### **Principal Components Analysis**

Principal Component Analysis (PCA) is a dimensionality reduction technique that transforms high-dimensional data into a smaller set of uncorrelated variables called principal components. It retains most of the original data's variability while reducing redundancy and multicollinearity. PCA is particularly useful in high-dimensional datasets, as it significantly reduces computational costs and enhances model efficiency by simplifying the input data without sacrificing critical information. 

The steps involved include:

1. Standardization: Features are normalized to have a mean of 0 and standard deviation of 1, ensuring they contribute equally to the analysis.

2. Covariance Matrix: A matrix is computed to capture the relationships (covariance) between all features.

3. Decomposition: The covariance matrix is decomposed into eigenvalues and eigenvectors. Eigenvectors represent directions of maximum variance (principal components), while eigenvalues measure the magnitude of variance in those directions.

4. Selecting Principal Components: Components are ranked by eigenvalues, and the top components explaining the desired variance , in our case we selected those who explained 97.5% of the variance. 

After doing the previous steps we calculated the **Kaiser's criterion**. It is a method used to determine the number of principal components to retain. According to this criterion, only components with eigenvalues greater than 1 are selected. The rationale is that an eigenvalue represents the amount of variance explained by a component, and values greater than 1 indicate that the component explains more variance than an individual original feature. We applied in two ways, covariance-based and correlation-based matrices. By systematically evaluating and optimizing these models, we aim to determine whether machine learning models can effectively address the forecasting challenges posed by the dataset.

The metrics selected for comparing the models are:

#### **Metrics for forecasting** 

##### 1. **Mean Absolute Error (MAE)**

MAE measures the average magnitude of the absolute errors between predictions ($y_{\text{pred}}$) and actual values ($y_{\text{true}}$).
$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \left| y_{\text{true}, i} - y_{\text{pred}, i} \right|
$$
Where:
- $n$: Number of data points.
- $y_{\text{true}, i}$: Actual value for observation $i$.
- $y_{\text{pred}, i}$: Predicted value for observation $i$. 
**Interpretation:**  
A lower MAE indicates better model performance. It is scale-dependent and does not consider the direction of errors (over- or under-prediction).
##### 2. **Root Mean Squared Error (RMSE)**

RMSE is the square root of the average squared differences between predicted and actual values, penalizing larger errors more than MAE.

$$
\text{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} \left( y_{\text{true}, i} - y_{\text{pred}, i} \right)^2}
$$

**Interpretation:**  
Similar to MAE but more sensitive to large deviations or outliers. A lower RMSE signifies better accuracy.

##### 3. **R-squared (R²)**

R², or the coefficient of determination, quantifies the proportion of the variance in the dependent variable explained by the independent variables.

$$
R^2 = 1 - \frac{\sum_{i=1}^{n} \left( y_{\text{true}, i} - y_{\text{pred}, i} \right)^2}{\sum_{i=1}^{n} \left( y_{\text{true}, i} - \bar{y}_{\text{true}} \right)^2}
$$

Where:
- $\bar{y}_{\text{true}}$: Mean of the actual values.

**Interpretation:**  
- $R^2 = 1$: Perfect prediction.
- $R^2 = 0$: Predictions are as good as using the mean of the actual values.
- $R^2 < 0$: Model performs worse than simply predicting the mean.

##### 4. **Mean Absolute Percentage Error (MAPE)**

MAPE calculates the average percentage error between predicted and actual values, offering a scale-independent interpretation.

$$
\text{MAPE} = \frac{100}{n} \sum_{i=1}^{n} \left| \frac{y_{\text{true}, i} - y_{\text{pred}, i}}{y_{\text{true}, i}} \right|
$$

**Interpretation:**  
Expressed as a percentage, MAPE provides intuitive interpretability. However, it can become undefined if any $y_{\text{true}, i} = 0$.

Long Short-Term Memory (LSTM) networks are a powerful innovation within machine learning, specifically tailored for sequential data. They emerged to address the limitations of traditional Recurrent Neural Networks (RNNs), particularly the vanishing gradient problem. This issue hampers the ability of standard RNNs to retain important long-term information during training, which is crucial for understanding sequences that unfold over extended periods.

### Core Components of LSTMs
- Input Gate: This mechanism determines what new data should be stored in the memory cell. It evaluates the incoming information and decides which elements are relevant to retain.
- Forget Gate: As its name suggests, this gate identifies and removes unnecessary or outdated information from the memory cell, allowing the model to prioritize more current and relevant data.
- Output Gate: This final gate regulates the output of information from the memory cell to either subsequent time steps or layers in the network, ensuring only meaningful insights are passed forward.

Why using LSTMs ?

- Temporal Dependencies: Stock prices often exhibit intricate patterns influenced by time. For instance, trends, cycles, and seasonal effects play a vital role in predicting future movements. LSTMs are designed to recognize these patterns by effectively connecting past and present data points.
- Handling Multivariate Data: Stock forecasting often requires analysis across multiple variables, such as individual stock prices, macroeconomic indicators, or global market trends. LSTMs can process such multidimensional data seamlessly, capturing relationships among variables that span across time.

# **Portfolio Optimization**
Throughout this project, we have utilized machine learning models to predict the close-to-close prices of various stocks. These predictions serve as the foundation for developing optimization algorithms designed to maximize the return on selected investments while minimizing associated risks. The proposed artificial intelligence model leverages predicted price trends and historical return data to make informed decisions on whether to buy, sell, or hold specific stocks and determines the optimal timing for executing these actions.

The optimization process is further constrained by three primary factors that reflect real-world limitations. First, an initial capital of €10,000 is available for investment. Second, there is a maximum limit of 100 stock purchases per day, ensuring practicality in trading volume. Finally, an operational cost of $1 USD per transaction is imposed, reflecting the financial impact of each trade. These constraints guide the design and implementation of the optimization algorithms.

To address this problem effectively, two principal approaches are considered. The first approach involves **exact search algorithms**, which aim to explore the solution space exhaustively to identify the optimal solution. These algorithms, such as linear programming or branch-and-bound, guarantee optimality but can be computationally intensive, particularly for large datasets or problems with intricate constraints. The second approach involves **metaheuristic algorithms**, which provide approximate solutions through heuristic techniques. Examples include genetic algorithms, simulated annealing, and particle swarm optimization. Metaheuristics are well-suited for large-scale or highly constrained problems, offering a practical trade-off between computational efficiency and solution quality.

Given the high computational complexity of the problem, we have opted for a metaheuristic approach, specifically focusing on single-objective genetic algorithms.

## **Genetic Algorithms**
<div style="text-align: center;">
  <img src="../images/ag.webp" style="width: 100%; max-width: 800px;"/>
</div>


**Genetic algorithms (GAs)** are optimization techniques inspired by the principles of natural selection and biological evolution. They are particularly useful for solving complex problems with large or poorly understood search spaces, where traditional optimization methods may struggle. By mimicking evolutionary processes, GAs iteratively refine a population of candidate solutions, moving closer to an optimal or near-optimal solution over successive generations.

The foundation of genetic algorithms lies in their representation of potential solutions, known as "chromosomes." These can be encoded as binary strings, numerical arrays, or other structures tailored to the specific problem. A fitness function evaluates each chromosome's quality, quantifying its ability to solve the given problem.

The evolutionary cycle of genetic algorithms consists of several core steps. First, the algorithm selects a subset of the population for reproduction based on their fitness scores. Selection methods, such as roulette wheel selection, tournament selection, or rank-based selection, ensure that higher-quality solutions have a better chance of contributing to the next generation. Next, selected chromosomes undergo crossover, where genetic material from two parent chromosomes is combined to produce offspring. This step enables exploration of new solution spaces by exchanging traits between parents.

Mutation, another critical step, introduces small random changes to the offspring's genetic material. This helps maintain diversity in the population and prevents the algorithm from converging prematurely to suboptimal solutions. Finally, a replacement strategy determines which chromosomes from the old population are replaced by the new offspring. Techniques like elitism ensure that the best solutions are retained across generations.

## **Chromosome representation**
In our implementation, each chromosome is represented as a 2D matrix where rows correspond to different assets (stocks) and columns represent days in the optimization horizon. Each element in the matrix specifies the number of units of a given stock bought or sold on a specific day. Positive values indicate purchases, negative values denote sales, and zeros signify no action. For example, consider the following simplified chromosome for three assets over five days:



\begin{bmatrix}
  10 & 0 & -5 & 0 & 0 \\
  0 & 20 & 0 & 0 & -10 \\
  0 & 0 & 0 & 15 & 0 \\
\end{bmatrix}


In this matrix: 
- Asset 1 (first row) starts with a purchase of 10 units on Day 1 and later sells 5 units on Day 3.
- Asset 2 (second row) shows a purchase of 20 units on Day 2, followed by a sale of 10 units on Day 5.
- Asset 3 (third row) records only one transaction, buying 15 units on Day 4.

Note that the chromosome must not exceed the available capital at any given time, and all stocks must be sold by the end of January.

## **Crossover**
The crossover operation is a process that combines the genetic material (chromosomes) of two parent solutions to create offspring. The idea is to produce new solutions that inherit **characteristics from both parents**, potentially combining their strengths and leading to better solutions over generations.In this particular implementation,the crossover method focuses on the most active assets within the parent portfolios, rather than swapping entire portfolios or randomly selecting parts of each chromosome. This approach targets assets that have experienced the highest levels of buying and selling activity.

To begin, the algorithm identifies the assets with the highest levels of activity in each parent portfolio. These are the assets that have been bought and sold the most. By focusing on these active assets, the crossover ensures that the offspring inherit the most relevant investment strategies from both parents. Once the most active assets have been identified, the method creates offspring by shifting investments between the two parent portfolios. The crossover leverages the most significant trades from one parent and applies them to the other parent's portfolio. This might involve buying or selling different assets based on the investment behaviors of the parents.

Finally, the crossover process undergoes iterative refinement. It is repeated over several iterations to ensure that the offspring portfolios meet specific validity criteria. The algorithm checks whether the adjusted portfolios remain feasible, ensuring that they maintain a realistic allocation of assets and do not exceed certain constraints.This iterative process continues until the offspring portfolios satisfy the necessary conditions. If there are no feasable solutions, the original parents´chromosomes are returned.An example is shown below : 


$$
\overset{\text{Parent 1}}{\begin{bmatrix}

## **Selection**
In genetic algorithms, the selection process determines which individuals (chromosomes) from the population will be chosen to reproduce and create offspring for the next generation. The aim is to favor individuals with higher fitness, ensuring that the best traits are passed on. There are several methods for selection, each with its advantages and trade-offs. Some common approaches include roulette wheel selection, rank selection, and tournament selection.

In our specific case, we have chosen tournament selection as the method for selecting parents from the population. This approach involves randomly selecting a subset of individuals from the population, called a "tournament," and then selecting the best individual from that subset based on their fitness. The tournament size can vary, but typically a small group of individuals is selected to compete against each other.

## **Mutation**

The mutation process is a critical operation in genetic algorithms, designed to introduce diversity and variability within the population of solutions. In our approach, mutation is applied to chromosomes with the aim of exploring new solutions while respecting the problem’s constraints. The mutation process consists of two primary steps: selecting mutation points and applying the mutation.

The first step is to **identify valid mutation points within the chromosome**. A mutation point consists of an asset and a day, where the asset represents the stock being traded and the day corresponds to the time when the trade action occurs. To ensure the mutation respects the problem's constraints, the algorithm checks the previous trades for the selected asset to confirm that there has been a valid purchase before the selected day and a corresponding sale after that day. Only those points that meet this condition are selected for mutation, ensuring that the changes adhere to the chronological order of stock transactions.

Once valid mutation points are selected, the mutation is applied by altering the stock trades at those points:

**1. Adjusting Existing Purchases and Sales:** For each selected mutation point, the algorithm adjusts the quantity of the stock held by reducing it, representing a partial sale of the stock. This reduction in stock is applied in a way that respects the earlier purchase and sale pattern, ensuring no inconsistency in the transaction history.

**2. Reinvesting the Capital:** The capital obtained from the sale of stocks is then used to purchase a new stock. The amount of money available for the new purchase is determined based on the sale proceeds. The algorithm calculates how many units of the new stock can be bought with the available funds. The chromosome is updated to reflect this new purchase, and the new asset is incorporated into the trading plan.

**3. Selling the New Stock:** After the new stock is purchased, the algorithm ensures that the stock is sold before the end of the investment period. A random day within the remaining days of the month is selected for the sale. This ensures that all stocks are sold by the end of the month, adhering to the problem's constraints.

Through these steps, the mutation process ensures that the newly generated chromosomes respect the constraints of the problem, such as the available capital and the requirement to sell all stocks by the end of the month. By introducing randomness in the mutation points and the actions taken, the algorithm explores different potential solutions, enhancing the search for an optimal portfolio. ## **Elitism**
Elitism is a mechanism in Genetic Algorithms (GAs) that ensures the best-performing individuals from one generation are carried over to the next without modification. This strategy helps preserve high-quality solutions and prevents the algorithm from losing optimal candidates due to stochastic effects like mutation or crossover.

The concept of elitism can be traced back to evolutionary biology, aligning with Lamarckian and Baldwinian theories of adaptation. Lamarckian evolution suggests that organisms pass on acquired characteristics to their offspring, implying a direct transfer of advantageous traits. In GAs, this would correspond to directly encoding learned or optimized features into the solution space. On the other hand, Baldwinian evolution emphasizes that adaptability, rather than acquired traits, is heritable. This perspective suggests that offspring do not inherit exact improvements but retain the capacity to adapt, akin to how mutation and crossover refine solutions in GAs.

## **Chromosome fitness evaluation : CARA Function**

In Genetic Algorithms, the evaluation of a chromosome is essential for determining the effectiveness of the portfolio configuration it represents. The goal is to assess how well the chromosome achieves the desired balance between risk and return. To do this, we calculate the portfolio's performance using the Certainty-Equivalent Risk Aversion (CARA) objective function, which incorporates both the portfolio's expected return and its associated risk.

The first step in the evaluation process is to calculate the portfolio weights based on the buy/sell quantities encoded in the chromosome. Each entry in the chromosome corresponds to a buy, sell, or hold decision for a particular asset on a specific day. These decisions are used to calculate the net inflows and outflows for each asset. For each asset, we sum the total purchases (negative values) and sales (positive values) based on the chromosome's decisions. We then normalize these values to obtain the portfolio weights, which represent the capital allocation for each asset. These weights are crucial because they determine how much capital is invested in each asset, which directly impacts both the portfolio's return and risk.

Once the portfolio weights are computed, we proceed to calculate the expected return of the portfolio. This is done by taking the weighted sum of the expected returns of each asset, where the weights are the proportions of capital allocated to each asset. The expected returns are calculated using historical data, as the mean return for each asset over a period. The resulting portfolio return provides an estimate of the potential profit that the portfolio is expected to generate.

The next step is to assess the portfolio's risk, which is represented by the variance of the portfolio's returns. Risk is quantified using the covariance matrix of asset returns, which captures how the returns of different assets move together over time. The portfolio's risk is calculated by multiplying the portfolio weights by the covariance matrix and then taking the dot product. This results in a measure of how much the portfolio's value is expected to fluctuate due to the correlation between the assets' returns.

Finally, we calculate the CARA objective function, which combines the portfolio's expected return and risk. The CARA function incorporates a risk aversion coefficient, denoted as $\gamma$. The objective function is designed to account for the fact that investors may prefer a more stable (lower-risk) portfolio, even at the cost of lower returns. The CARA value is computed by subtracting the risk term (scaled by $\gamma$) from the portfolio's expected return. A higher CARA value indicates a better portfolio configuration, as it represents a higher return for a given level of risk, aligning with the investor's preferences.

The CARA value is then calculated as:

$$ \text{CARA Value} = R_p - \frac{\gamma}{2} \cdot \sigma_p^2 $$

Where:
- $R_p$ represents the portfolio return.
- $\sigma_p^2$ represents the portfolio risk (variance).
- $\gamma$ is the risk aversion parameter.
